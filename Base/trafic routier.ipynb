## Affichage plus joli

**Énoncé** Peu d'êtres humains aiment lire des séries de nombres. Un affichage plus joli sera aussi plus pratique.

Programmez une fonction `dessiner_file` qui, une file lui étant donnée, renverra une chaîne de caractères obtenue en remplaçant chaque nombre positif par un `"🚗"` suivi du nombre entre parenthèses et les `0` par des `"🌳"`. Et comme `"🚗"` est dirigée vers la gauche, il faudra que la tête de la file soit à gauche. Séparez deux cases consécutives par un caractère espace `" "`.

**Exemple** Notre file `voie` était affichée ainsi :
```
(fin) 5->0->3->1->0->0->2 (début)
```
si on passe `voie` en paramètre à `dessiner_file`, cette fonction renverra la chaîne de caractères :
```
'🚗(2) 🌳 🌳 🚗(1) 🚗(3) 🌳 🚗(5)'
```

**Bonus** Faites en sorte que la file passée en paramètre ne soit pas détruite.

**Indication** Vous pouvez traiter `"🌳"` et `"🚗"` comme n'importe quels caractères Python, en les entourant de guillemets :

````python
print("🚗"+"🌳")
````

````python
# votre code ici
def dessiner_file(F):
    
    affichage = []
    index = 0
    
    for voiture in F:
        if voiture == 0:
            affichage.append("🌳")
        
        elif voiture > 0:
            affichage.append("🚗(" + str(voiture) + ")")
            
        else:
            affichage.append("🪨")
        
        index += 1
        
    affichage = " ".join(affichage)
            
    return affichage
````

````python
dessiner_file(voie)
````

## Création aléatoire

**Énoncé** Pour faire des tests, il faut des files. Plutôt que de les créer à la main, ce qui peut être très long, programmez une fonction `creer_liste_aleatoire` qui prend en paramètre un nombre entier naturel `n` et qui renvoie une file de `n` nombres entiers naturels tirés au hasard entre `0` et `n`.

**Bonus 1** Faites en sorte qu'environ une case sur deux soit vide (ici, "environ une case sur deux" signifie : une chance sur deux que chaque case créée soit vide).

**Bonus 2** Faites en sorte qu'à part les 0, aucun nombre ne se répète dans une file.

**Indication** Pour tirer au hasard un nombre entier aléatoire compris `2` et `5`, on peut utiliser :

````python
import random
random.randint(2,5)
````

````python
# votre code ici
def creer_liste_aleatoire(n):
    
    liste = []
    nb = []
    
    while len(liste) != n:
        chance = random.randint(1,2)
        
        if chance == 1:
            vehicule = random.randint(1,n)
        
            if vehicule not in nb:
                liste.append(vehicule)
                nb.append(vehicule)
                
        elif chance == 2: 
            liste.append(0)
            
    
    return liste
        
````

````python
test = creer_liste_aleatoire(9)
dessiner_file(test)
````

## Avancer jusqu'à...

**Énoncé** On considère `F` une file et `m` un nombre entier naturel. Programmez une fonction Python `avancer_jusque` qui prend comme paramètres `F` et `m` et qui fait avancer d'une case toutes les voitures situées sur les `m` premières cases de la file. Cette fonction doit modifier la file passée en paramètre. 

Si une voiture est placée en tête de la file avant application de la fonction, elle avance et sort de la file. On suppose que `m` est compris entre 1 et le nombre d'éléments de la file. La case numéro `m` est laissée vide.

**Exemple** Avec la file `(fin) 5->0->3->1->0->0->2 (début)`, si `m` est égal à 4, on fait avancer les voitures des 4 premières cases et on obtient `(fin) 5->0->3->0->1->0->0 (début)`. Si vous préférez : `'🚗(2) 🌳 🌳 🚗(1) 🚗(3) 🌳 🚗(5)'` devient `'🌳 🌳 🚗(1) 🌳 🚗(3) 🌳 🚗(5)'`

````python
# votre code ici
def avancer_jusque(F, m):
    
    part = F[:m]
    
    defiler(part)
    enfiler(part, 0)
    
    F[:m] = part
````

## Avancer après...

**Énoncé** Même problème que ci-dessus, mais la fonction `avancer_apres` va faire avancer (si possible), toutes les voitures situées sur la case numéro `m` et après, d'un cran. On complètera la dernière case avec un zéro.

**Exemple 1** Avec la file `(fin) 5->0->3->1->0->0->2 (début)`, si `m` est égal à 4, on fait avancer les voitures de cette case et des suivantes et on obtient `(fin) 0->5->0->3->1->0->2 (début)`. Si vous préférez : `'🚗(2) 🌳 🌳 🚗(1) 🚗(3) 🌳 🚗(5)'` devient `'🚗(2) 🌳 🚗(1) 🚗(3) 🌳 🚗(5) 🌳'`

**Exemple 2** Toujours avec la file `(fin) 5->0->3->1->0->0->2 (début)`, si `m` est égal à 5, on ne peut pas faire avancer la voiture `🚗(3)` située case 5 car elle est bloquée par `🚗(1)`.

````python
# votre code ici
def avancer_apres(F, m):
    
    part = F[m-2:]
    
    defiler(part)
    enfiler(part, 0)
    
    F[m-2:] = part
````

## Avancer rapidement

**Énoncé** Parfois, une voiture est beaucoup plus rapide que les autres. Cela peut être un camion de pompier, une ambulance... ou un chauffard. Programmez une fonction `avancer_rapide` qui prend deux paramètres : une file `F` de voitures et un nombre entier `m` qui est compris entre 1 et le nombre de cases de `F`. Cette fonction doit faire avancer la voiture située dans la case numéro `m` (s'il y en a une) d'autant de cases que possible. Les autres voitures ne bougent pas et la case laissée par la voiture rapide est laissée vide. Attention, on ne peut pas doubler !

**Exemple 1** Avec la file `(fin) 5->0->3->1->0->0->2 (début)`, si `m` est égal à 4, on fait avancer la voiture `🚗(1)` de deux cases et on obtient `(fin) 5->0->3->0->0->1->2 (début)`. Si vous préférez : `'🚗(2) 🌳 🌳 🚗(1) 🚗(3) 🌳 🚗(5)'` devient `'🚗(2) 🚗(1) 🌳 🌳 🚗(3) 🌳 🚗(5)'`

**Exemple 2** Toujours avec la file `(fin) 5->0->3->1->0->0->2 (début)`, si `m` est égal à 5, on ne peut pas faire avancer la voiture `🚗(3)` située case 5 car elle est bloquée par `🚗(1)`.

````python
# votre code ici
def avancer_rapide(F, m):
    
    part = F[:m]
    index = m - 1
    
    while index > 0 and part[index - 1] == 0:
        part[index - 1] = part[index]
        part[index] = 0
        index -= 1
        
    F[:m] = part
````

````python
test = creer_liste_aleatoire(9)
print(test)
avancer_rapide(test, 4)
dessiner_file(test)
````

## Obstacle

**Énoncé** Lorsqu'il y a un obstacle, la route se dégage avant l'obstacle et un embouteillage se crée à partir de l'obstacle. Programmez une fonction `obstacle` qui prend deux paramètres : une file de voitures `F` et un nombre entier `m` compris entre 1 et le nombre d'éléments de la file. La fonction doit :

- faire sortir de la file toutes voitures situées avant la position `m` ;
- faire en sorte que toutes les voitures situées à la position `m` ou après se retrouvent tassées à partir de la position `m`, c'est-à-dire situées uniquement en `m`, `m+1`, éventuellement `m+2` etc, sans case vide entre deux voitures.

On suppose que les secours ont bloqué l'entrée dans la file, donc la file obtenue doit garder la même longueur et se terminer éventuellement par des zéros.

**Exemple 1** Avec la file `'🚗(2) 🌳 🌳 🚗(1) 🚗(3) 🌳 🚗(5)'`, si `m` est égal à 4, on vide les 3 premières cases. Ensuite, seule la voiture `🚗(5)` peut avancer d'un cran. La file  devient `'🌳 🌳 🌳 🚗(1) 🚗(3) 🚗(5) 🌳 🌳'`

**Exemple 2** Même file, mais avec `m` égal à 2. Cette fois, on vide la case 1, les trois voitures suivantes peuvent avancer pour se tasser à partir de la case 2 et on obtient : `'🌳 🚗(1) 🚗(3) 🚗(5) 🌳 🌳 🌳'`

````python
# votre code ici
def obstacle(F, m):
    
    """
    On vide la parti avant l'obstacle
    """
    part = F[:m-1]
    
    for i in range(len(part)):
        defiler(part)
        enfiler(part, 0)
            
    F[:m-1] = part
    
    """
    On met toute les voitures dans l'ordre apès l'obstacle
    """
    part2 = F[m-1:]
    voitures = [x for x in part2 if x != 0]
    
    while len(voitures) < len(part2):
        voitures += [0]
    
    for index in range(len(part2)):
        part2[index] = voitures[index]
        
    F[m-1:] = part2
    
````

````python
test = creer_liste_aleatoire(9)
print(test)
obstacle(test, 5)
dessiner_file(test)
````
